```python
import pandas as pd
from pathlib import Path

# 🔧 Configuration
csv_folder = Path(r"C:\Users\csofios\Desktop\LongTermCSO\SWMM\pump status(35-45mph)")  # Set your folder path here
target_pumps = {"MDOT_PumpI94Cadieux", "MDOT_PumpI94Whittier", "MDOT_PumpI94Outer","MDOT_PumpI94Dickerson","MDOT_PumpI94Cadillac","FVSanPump1","FVSanPump2","FVSanPump3",
               "FVSanPump4","FischerPump1","FischerPump2","CC_Dewater1","CC_Dewater2","CC_Dewater3","CC_Dewater4","PUMP56@1759-1760","PUMP51@7004-7000","NeffPS_StormP7","NeffPS_StormP3","NeffPS_StormP2","NeffPS_StormP1",
                "NeffPS_SanP6","NeffPS_SanP5","NeffPS_SanP4","GPS-CookRoadPS","Kerby1","Kerby2","Kerby3","Kerby4","Kerby5","BH_SanPump1","BH_SanPump2",
               "MDOT_PumpI94Seneca","MDOT_PumpI94Conner","MDOT_PumpI94Harper","HW_PumptoBlackMarsh","BH_StormPump1","BH_StormPump2","BH_StormPump3","BH_StormPump4",
               "CON-PMP-1","CON-PMP-2","CON-PMP-3","CON-PMP-4","CON-PMP-5","CON-PMP-6","CON-PMP-7","CON-PMP-8","CON-PMP-9","CON-PMP-10","CON-PMP-11","CON-PMP-12",
               "FRE-PMP-1","FRE-PMP-2","FRE-PMP-3","FRE-PMP-4","FRE-PMP-5","FRE-PMP-6","FRE-PMP-7","FRE-PMP-8","FRE-PMP-9","FRE-PMP-10"}  # Pump names you want to activate

# 🔁 Iterate over all CSV files in the folder
for csv_file in csv_folder.glob("*.csv"):
    try:
        df = pd.read_csv(csv_file)

        if "Name" in df.columns and "Status" in df.columns:
            # Update status only if pump name is in target_pumps and status is currently 0
            mask = (df["Name"].isin(target_pumps)) & (df["Status"] == 0)
            df.loc[mask, "Status"] = 1

            # Save back to the same file
            df.to_csv(csv_file, index=False)
            print(f"✅ Updated: {csv_file.name} ({mask.sum()} pumps changed)")
        else:
            print(f"⚠️ Skipped (missing columns): {csv_file.name}")
    except Exception as e:
        print(f"❌ Failed to process {csv_file.name}: {e}")
```
