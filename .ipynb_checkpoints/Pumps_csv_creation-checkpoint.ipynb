```python
import pandas as pd
import os

# Path where your 100 CSV files are stored
folder_path = r"C:\Users\csofios\Desktop\LongTermCSO\SWMM\pump status(35-45mph)"

# Start with the first file to create base DataFrame
base_df = pd.read_csv(os.path.join(folder_path, 'pump_status_1.csv'))
merged_df = base_df.iloc[:, :2].copy()  # Keep only first two columns (e.g., Time and ID)
merged_df['Status_1'] = base_df['Status']  # Add Status_1

# Loop through pump_status_2.csv to pump_status_100.csv
for i in range(2, 101):
    file_name = f'pump_status_{i}.csv'
    file_path = os.path.join(folder_path, file_name)
    
    df = pd.read_csv(file_path)
    merged_df[f'Status_{i}'] = df['Status']  # Add the Status column with a new name

# Save final merged DataFrame
output_path = os.path.join(folder_path, 'pumps_35-45.csv')
merged_df.to_csv(output_path, index=False)

print(f"Merged file saved to: {output_path}")
```
